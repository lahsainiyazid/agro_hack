```python
import numpy as np 
import pandas as pd 
import sklearn 
import setuptools 
import emlearn 
```

```python
df=pd.read_csv('/home/yazid/agro/datasets/weatherAUS.csv')
```

```python
df.head()
```

```python
df['Date']=pd.to_datetime(df['Date'])
df['Month']=df['Date'].dt.month
df['month_cos']=np.cos(2*np.pi*df['Month']/12)
df['month_sin']=np.sin(2*np.pi*df['Month']/12)
```

```python
cols_to_keep=['month_sin','month_cos','Humidity9am','Humidity3pm','Pressure9am','Pressure3pm','RainTomorrow']
```

```python
df=df[cols_to_keep]
```

```python
df['delta_hum']=df['Humidity3pm']-df['Humidity9am']
df['delta_pres']=df['Pressure9am']-df['Pressure3pm']
```

Let's treat missing values:


```python
(df.isna().sum()/len(df))*100
```

```python
df.dropna(subset="RainTomorrow",inplace=True)
df.dropna(subset="Humidity9am",inplace=True)
df.dropna(subset="Humidity3pm",inplace=True)
df.dropna(subset="delta_hum",inplace=True)
```

```python
df.isna().sum()/len(df) *100
#Now we have to treat the pressure:
```
